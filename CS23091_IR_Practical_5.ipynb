```python
# Step 1: Import libraries
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.metrics.pairwise import cosine_similarity
```

```python
# Step 2: Define two documents
doc1 = "Machine learning is very powerful"
doc2 = "Machine learning is part of artificial intelligence"
```

```python
documents = [doc1, doc2]
```

```python
# Step 3: Convert text into vectors (Bag of Words)
vectorizer = CountVectorizer(stop_words='english')
vectors = vectorizer.fit_transform(documents)
```

```python
# Step 4: Display Vocabulary
print("Vocabulary:", vectorizer.get_feature_names_out())
```

```python
# Step 5: Convert sparse matrix to array
vector_array = vectors.toarray()
print("\nDocument Vectors:")
print(vector_array)
```

```python
# Step 6: Compute Cosine Similarity
similarity = cosine_similarity(vector_array[0].reshape(1, -1),
                               vector_array[1].reshape(1, -1))
```

```python
print("\nCosine Similarity between Doc1 and Doc2:", similarity[0][0])
```
